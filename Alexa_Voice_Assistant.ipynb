# 3:Record Audio From Microphone----

```python
import sounddevice as sd # help to record audio
from  scipy.io.wavfile import write # help to store audio
```

     record_audio() : It will Record audio from microphone for 5 seconds, at speech-quality sample rate, using one audio channel.

```python
def record_audio(filename = 'input2.wav',duration = 5,fs = 48000):
    print(" Listening")

    recording = sd.rec(int(duration*fs),samplerate=fs,channels=1)
    sd.wait()
    write(filename,fs,recording)
    print("Audio recording")
```

```python
record_audio(fs=48000)
```

```python
'''

If a .wav file has no audio, it usually means Python selected the wrong microphone & as a result, the recorded file contains no voice.

'''
```

```python
sd.query_devices() # List all your availabel devices for your system
```

```python
sd.default.device = (1,3) # set the correct physical microphone for recording...
```

```python
record_audio(fs = 48000)
```

# 6. Prepare training data to train alexa---

```python
'''

Alexa (or Alexa-like assistants) does not understand language naturally like humans.

    It learns by identifying patterns from data and examples..
    Without training, Alexa would:

    Guess randomly
    Give inaccurate or unreliable responses 😬



'''
```

```python
import pandas as pd
```

```python
alexa_df = pd.read_csv("alexa_data.csv")
```

```python
alexa_df.head(20)
```

```python
alexa_df.shape
```

```python
'''
Neural networks can only work with numbers, so text sentences must be converted into numerical form before training..

'''
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
vectorizer = TfidfVectorizer()
```

```python

X = vectorizer.fit_transform(alexa_df['prompt'])
```

```python
X.shape
```

```python
X.toarray()
```

```python
X.toarray()[0]
```

# 7 Train a simple deep learning intent model ----

```python
from sklearn.neural_network import MLPClassifier
```

```python
'''
MLP = Multi-Layer Perceptron
In simple words:
👉 It’s a basic neural network that learns patterns in data to make decisions.

'''
```

```python
intent_model = MLPClassifier(hidden_layer_sizes=(50,25),max_iter=500)
```

```python
intent_model.fit(X,alexa_df['intent'])
```

```python
def predict_intent(text):
    X_text = vectorizer.transform([text])
    intent = intent_model.predict(X_text)[0]
    print("Detected intent:", intent)
    return intent
```

```python
predict_intent("open google")
```

```python
predict_intent("i want to eat food and oder it by swiggy")
```

```python
predict_intent("open youtube")
```

```python
predict_intent("i want to buy butter please add it to my shopping list")
```
