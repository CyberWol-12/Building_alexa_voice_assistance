# 3:Record Audio From Microphone----

```python
import sounddevice as sd # help to record audio
from  scipy.io.wavfile import write # help to store audio
```

     record_audio() : It will Record audio from microphone for 5 seconds, at speech-quality sample rate, using one audio channel.

```python
def record_audio(filename = 'input2.wav',duration = 5,fs = 48000):
    print(" Listening")

    recording = sd.rec(int(duration*fs),samplerate=fs,channels=1)
    sd.wait()
    write(filename,fs,recording)
    print("Audio recording")
```

```python
record_audio(fs=48000)
```

```python
'''

If a .wav file has no audio, it usually means Python selected the wrong microphone & as a result, the recorded file contains no voice.

'''
```

```python
sd.query_devices() # List all your availabel devices for your system
```

```python
sd.default.device = (1,3) # set the correct physical microphone for recording...
```

```python
record_audio(fs = 48000)
```
